```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False,dtype = int)
encoded = encoder.fit_transform(df[["Gender", "Dependents"]])
df[encoder.get_feature_names_out()] = encoded
df.head()
```

```python
df = df.drop(["Gender", "Dependents"], axis=1)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numerical_col = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History"
]

df[numerical_col] = scaler.fit_transform(df[numerical_col])

df.head()
```
